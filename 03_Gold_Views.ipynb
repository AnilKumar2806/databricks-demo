```python
%sql
--- Monthly Sales
-- DROP TABLE IF EXISTS dev.demo.gold_monthly_sales;
CREATE OR REPLACE VIEW dev.demo.gold_monthly_sales AS
SELECT
    sales_month,
    SUM(sales_amount) total_sales,
    COUNT(*) total_orders
FROM dev.demo.fact_sales
GROUP BY sales_month;
```

```python
%sql
select * from dev.demo.gold_monthly_sales
```

```python
%sql
--- Customer Revenue
-- DROP TABLE IF EXISTS dev.demo.gold_customer_revenue;
CREATE OR REPLACE VIEW dev.demo.gold_customer_revenue AS
SELECT
    customer_id,
    customer_name,
    COUNT(*) total_orders,
    SUM(sales_amount) total_revenue,
    round(AVG(sales_amount),2) avg_order_value,
    MAX(sales_amount) largest_order,
    MIN(sales_amount) smallest_order
FROM dev.demo.fact_sales
GROUP BY customer_id, customer_name;
```

```python
%sql
select * from dev.demo.gold_customer_revenue
```

```python
%sql
--- Product Revenue
CREATE OR REPLACE VIEW dev.demo.gold_product_revenue AS
SELECT
    product_name,
    category_name,
    SUM(sales_amount) total_revenue,
    SUM(quantity) total_quantity
FROM dev.demo.fact_sales
GROUP BY product_name, category_name;
```

```python
%sql
--- Top Customers
CREATE OR REPLACE VIEW dev.demo.gold_top_customers AS
SELECT
    customer_name,
    SUM(sales_amount) AS total_revenue
FROM dev.demo.fact_sales
GROUP BY customer_name
ORDER BY total_revenue DESC;
```

```python
%sql
--- Top Products
CREATE OR REPLACE VIEW dev.demo.gold_top_products AS
SELECT
    product_name,
    SUM(sales_amount) AS total_revenue
FROM dev.demo.fact_sales
GROUP BY product_name
ORDER BY total_revenue DESC;
```
